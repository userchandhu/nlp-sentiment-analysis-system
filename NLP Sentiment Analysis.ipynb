```python
# importing libraries
import nltk 
import pandas as pd
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import MultinomialNB
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.metrics import accuracy_score, classification_report
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
# loading and make copy of data 
data = pd.read_csv(r"C:\Users\Prem\Desktop\ML projects\IMDB_dataset.csv")
df = data
```

```python
# Handling duplicates
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
# text cleaning
stop_words = set(stopwords.words('english'))

def clean_text(text):
    tokens = word_tokenize(text.lower())
    words = [ word for word in tokens if word.isalpha() ]
    return ' '.join(words)

df['review'] = df['review'].apply(clean_text)
```

```python
# splitting features and labels
x = df['review']
y = df['sentiment']
```

```python
# train test split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
# vectorization
vectorizer = TfidfVectorizer(ngram_range=(1,2))
x_train = vectorizer.fit_transform(x_train)
x_test = vectorizer.transform(x_test)
```

```python
# model building
nb_model = MultinomialNB()
nb_model.fit(x_train,y_train)
```

```python
# prediction
y_pred = nb_model.predict(x_test)
```

```python
# evaluation
acc = accuracy_score(y_test,y_pred)
report = classification_report(y_test,y_pred)

print('accuracy_score:',acc)
print('classification_report:',report)
```

```python
# prediction function
def predict_sentiment(text):
    text = clean_text(text)
    vector = vectorizer.transform([text])
    prediction = nb_model.predict(vector)
    return prediction[0]
```

```python
# testing 
print(predict_sentiment(' this is very intresting and good movie'))
print(predict_sentiment(' this is very bad movie'))
```

```python
# building model
lr_model = LogisticRegression(class_weight = 'balanced')
lr_model.fit(x_train, y_train)
```

```python
# prediction
y_pred = lr_model.predict(x_test)
```

```python
# evaluation
acc = accuracy_score(y_test,y_pred)
report = classification_report(y_test,y_pred)

print('accuracy_score:',acc)
print('classification_report:',report)
```

```python
# prediction function
def pred_sentiment(text):
    text = clean_text(text)
    vectors = vectorizer.transform([text])
    prediction = lr_model.predict(vectors)
    return prediction[0]
```

```python
print(pred_sentiment(' this is a good time for relaxing ' ))
print(pred_sentiment(' this is very bad movie, and too boring ' ))
print(pred_sentiment(" i don't like the party " ))
```
